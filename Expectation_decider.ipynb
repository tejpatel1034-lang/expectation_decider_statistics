# 1. Understanding the Basics

## What is Probability?

Probability is a measure of how likely an event is to occur.

It ranges from 0 to 1:

- 0 → Impossible event
- 0.25 → Low chance
- 0.50 → 50-50 chance
- 0.75 → High chance
- 1 → Certain event

Formula:

P(E) = Favourable Outcomes / Total Outcomes

## Key Probability Terminology

- Experiment: A process that produces an outcome.
- Outcome: A possible result of an experiment.
- Sample Space: The set of all possible outcomes.
- Event: A specific outcome or group of outcomes of interest.
- Favourable Outcome: An outcome that satisfies the condition of an event.
- Probability: The numerical measure of the chance that an event occurs.

### Probability Events from the Dataset

We consider the following events:

1. Event A: A randomly selected student passes the final exam.
2. Event B: A randomly selected student has attendance greater than 80%.
3. Event C: A randomly selected student participates in group discussion.

#### 1. Event A: A randomly selected student passes the final exam.

```python
total_students=len(df)

pass_students=(df["final_exam_pass"]=="Pass").sum()

p_pass=pass_students / total_students 

print("p(pass)=", p_pass)
```

#### 2. Event B: A randomly selected student has attendance greater than 80%.

```python
high_attendance = (df["attendance"] > 80).sum()
p_high_attendance = high_attendance / total_students
print("P(Attendance > 80%)=",p_high_attendance)
```

#### 3. Event C: A randomly selected student participates in group discussion.

```python
group_discussion_yes= (df["group_discussion"]== "yes").sum()
p_group_discussion = group_discussion_yes / total_students
print("P(Group Discussion = Yes)=", p_group_discussion)
```

#### Percenatage

```python
print("p(pass)=", round(p_pass*100,2),"%")
print("P(Attendance > 80%)=",round(p_high_attendance *100,2),"%")
print("P(Group Discussion = Yes)=", round(p_group_discussion * 100,2),"%")
```

# 2. Types of Events

## Empirical Probability

Empirical probability is calculated using actual observed data or past observations.

Formula:

P(E) = Number of times event occurs / Total number of observations

we will calculate the empirical probability of a student passing the final exam.

```python
pass_count = (df["final_exam_pass"] == "Pass").sum()
total_students = len(df)

empirical_probability = pass_count / total_students

print("Number of students who passed:", pass_count)
print("Total students:", total_students)
print("Empirical Probability of Passing:", empirical_probability)
print("Empirical Probability (%):", round(empirical_probability * 100, 2), "%")
```

## Theoretical Probability

Theoretical probability is the probability of an event calculated based on the possible outcomes, without actually performing the experiment.

### Formula

$$
P(E)=\frac{\text{Number of Favorable Outcomes}}{\text{Total Number of Possible Outcomes}}
$$

# 3. Random Variable & Probability Distribution

Let X be the number of students who pass the final exam out of 3 randomly selected students.

Therefore, the possible values of X are:

X = 0, 1, 2, 3

```python
from math import comb

n = 3
p = empirical_probability
q = 1 - p

x_values = [0, 1, 2, 3]

probabilities = []

for k in x_values:
    prob = comb(n, k) * (p ** k) * (q ** (n - k))
    probabilities.append(prob)

distribution = pd.DataFrame({
    "X (Number of Pass Students)": x_values,
    "P(X)": probabilities
})

distribution
```

#### Check Probabilty total

```python
distribution["P(X)"].sum()
```

### Mean of the Random Variable

The mean of a discrete random variable is calculated as:

E(X) = Σ [x × P(X = x)]

It represents the expected number of students who will pass out of 3 randomly selected students.

```python
mean_x = (distribution["X (Number of Pass Students)"] * distribution["P(X)"]).sum()

print("Mean (Expected Number of Pass Students):", mean_x)
```

### Variance of the Random Variable

Variance measures how much the values of the random variable vary around its mean.

Formula:

Var(X) = Σ [ (x - μ)² × P(X = x) ]

```python
variance_x = (
    (
        distribution["X (Number of Pass Students)"] - mean_x
    ) ** 2
    * distribution["P(X)"]
).sum()

print("Variance:", variance_x)
```

# 4. Venn Diagram

Let:

- A = Students who study for more than 10 hours
- B = Students whose attendance is greater than 80%
- C = Overlap showing students who satisfy both conditions.

The Venn diagram will show:

- A only
- B only
- C = A ∩ B (both conditions)
- Neither A nor B

```python
A = df["study_hours"] > 10
B = df["attendance"] > 80

A_only = (A & ~B).sum()
B_only = (~A & B).sum()
intersection = (A & B).sum()
neither = (~A & ~B).sum()

print("A only:", A_only)
print("B only:", B_only)
print("C = A ∩ B (both condtition):", intersection)
print("Neither:", neither)
```

#### Check total

```python
A_only+B_only+intersection+neither
```

# 5. Contingency Table & Probability

A contingency table shows the relationship between two categorical variables.

Here we compare:
- Group Discussion: Yes / No
- Final Exam: Pass / Fail

```python
contingency_table = pd.crosstab(
    df["group_discussion"],
    df["final_exam_pass"]
)

contingency_table
```

## Joint Probability

Joint probability is the probability that two events occur together.

Formula:

P(A ∩ B) = Number of observations satisfying both events / Total observations

```python
joint_probability = contingency_table / len(df)

joint_probability
```

## Marginal Probability

Marginal probability is the probability of one variable without considering the other variable.

```python
marginal_group = df["group_discussion"].value_counts(normalize=True)
marginal_pass = df["final_exam_pass"].value_counts(normalize=True)
```

```python
print("Marginal Probability - Group Discussion")
print(marginal_group)
```

```python

print("\nMarginal Probability - Final Exam")
print(marginal_pass)
```

## Conditional Probability

Conditional probability is the probability of an event occurring given that another event has already occurred.

For example:

P(Pass | Group Discussion = Yes)

```python
p_pass_given_discussion = (
    df[df["group_discussion"] == "Yes"]["final_exam_pass"]
    .eq("Pass")
    .mean()
)

print(
    "P(Pass | Group Discussion = Yes) =",
    p_pass_given_discussion
)
```

# 6. Understanding Relationships

The relationship between Group Discussion and Final Exam Pass is analyzed using conditional probability.

If the probability of passing changes depending on whether a student participates in group discussion, the two events may be dependent.

If the probabilities remain the same, the events may be independent.

```python
p_pass = (df["final_exam_pass"] == "Pass").mean()
```

```python
p_pass_given_yes = (
    df[df["group_discussion"] == "Yes"]["final_exam_pass"]
    .eq("Pass")
    .mean()
)
```

```python
p_pass_given_no = (
    df[df["group_discussion"] == "No"]["final_exam_pass"]
    .eq("Pass")
    .mean()
)
```

```python
print("P(Pass) =", p_pass)
print("P(Pass | Group Discussion = Yes) =", p_pass_given_yes)
print("P(Pass | Group Discussion = No) =", p_pass_given_no)
```

## Relationship Interpretation

If P(Pass | Group Discussion = Yes) is different from P(Pass), this suggests that the two events are dependent.

If P(Pass | Group Discussion = Yes) is approximately equal to P(Pass), this suggests independence.

Group Discussion = Yes and Final Exam = Pass are not mutually exclusive because both events can occur for the same student.

# 7. Bayes Theorem

Bayes Theorem is used to calculate the probability of an event based on prior information.

Formula:

P(A|B) = [P(B|A) × P(A)] / P(B)

Given:

P(H|Pass) = 0.70
P(H|Fail) = 0.40
P(H) = 0.60

Find:

P(Pass|H)

```python
p_h_given_pass = 0.70
p_h_given_fail = 0.40
p_h = 0.60
```

```python
p_pass = (p_h - p_h_given_fail) / (p_h_given_pass - p_h_given_fail)
p_fail = 1 - p_pass
```

```python
p_pass_given_h = (p_h_given_pass * p_pass) / p_h
```

```python
print("P(Pass) =", p_pass)
print("P(Fail) =", p_fail)
print("P(Pass | H) =", p_pass_given_h)
print("P(Pass | H) (%) =", round(p_pass_given_h * 100, 2), "%")
```

## Interpretation

Based on the given probabilities and Bayes Theorem, the probability of Pass given H is 75%.

Therefore:

P(Pass | H) = 0.75
